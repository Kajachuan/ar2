```python
import numpy as np
import matplotlib.pyplot as plt

env_name = 'roundabout-v0'
root = f'{env_name}-PPO'

data = np.load(f"{root}/logs/evaluations.npz")
timesteps    = data["timesteps"]
mean_rewards = data["results"].mean(axis=1)
std_rewards  = data["results"].std(axis=1, ddof=1)

plt.figure(figsize=(10, 5))
plt.plot(timesteps, mean_rewards, label="Mean Reward")
plt.fill_between(
    timesteps,
    mean_rewards - std_rewards,
    mean_rewards + std_rewards,
    alpha=0.3,
    label="±1 std"
)
plt.xlabel("Timesteps")
plt.ylabel("Mean Reward")
plt.title("Curva de Convergencia — PPO Roundabout")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{root}/logs/convergence_curve.png", dpi=150)
plt.show()
```
